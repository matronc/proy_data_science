## 1. Parámetros configurables

Centralizar aquí los parámetros evita valores 'mágicos' dispersos en el notebook.

```python
# ── Parámetros ─────────────────────────────────────────────────────────────
RUTA_DATOS   = "../data/resumen_ventas_preparado.csv"
CARPETA_OUT  = "../outputs"
COL_CATEGORIA = "Categoria"
COL_PRECIO    = "Precio_Medio"
COL_VENTAS    = "Total_Ventas_Acumuladas"
```

## 2. Importación de librerías

```python
import sys
import os

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Agregar el directorio scripts al path para importar utilidades
sys.path.append(os.path.join("..", "scripts"))
from utils import cargar_csv, limpiar_dataframe, convertir_tipos, resumen_estadistico, ventas_por_categoria, grafico_barras

# Estilo de visualización
sns.set_theme(style="whitegrid")
print("Librerías importadas correctamente.")
```

## 3. Carga de datos

```python
# Cargar el CSV con la función reutilizable de utils.py
df = cargar_csv(RUTA_DATOS)
df.head()
```

```python
# Vista general del dataset
print(f"Dimensiones: {df.shape}")
print("\nTipos de datos:")
display(df.dtypes)
print("\nValores nulos por columna:")
display(df.isnull().sum())
```

## 4. Limpieza de datos

```python
# Eliminar duplicados y filas completamente vacías
df = limpiar_dataframe(df)

# Asegurar que las columnas numéricas tienen el tipo correcto
df = convertir_tipos(df, [COL_PRECIO, COL_VENTAS])

df.head()
```

## 5. Análisis exploratorio

```python
# Estadísticas descriptivas de las ventas totales
stats = resumen_estadistico(df, COL_VENTAS)
print("Estadísticas — Total Ventas Acumuladas")
for k, v in stats.items():
    print(f"  {k:10s}: {v}")
```

```python
# Ventas agrupadas por categoría
resumen = ventas_por_categoria(df, COL_CATEGORIA, COL_VENTAS)
display(resumen)
```

```python
# Precio medio por categoría
precio_por_cat = ventas_por_categoria(df, COL_CATEGORIA, COL_PRECIO)
display(precio_por_cat)
```

## 6. Visualización

```python
# Gráfico de barras: total de ventas por categoría
ruta_img = grafico_barras(
    resumen,
    col_x=COL_CATEGORIA,
    col_y="total",
    titulo="Total Ventas por Categoría",
    carpeta_salida=CARPETA_OUT
)
print(f"Imagen guardada en: {ruta_img}")
```

```python
# Gráfico adicional: precio medio por categoría
grafico_barras(
    precio_por_cat,
    col_x=COL_CATEGORIA,
    col_y="total",
    titulo="Precio Medio por Categoría",
    carpeta_salida=CARPETA_OUT
);
```

## 7. Conclusiones

- La categoría **Tech** concentra el mayor volumen de ventas acumuladas.
- El precio medio de **Tech** duplica al de **Hogar**, lo que explica la diferencia en ingresos.
- Los datos están limpios (sin duplicados ni nulos), por lo que el pipeline es estable.
